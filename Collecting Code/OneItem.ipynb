```python
import pyodbc
import pandas as pd
from datetime import datetime

def get_sql_connection():
    """ایجاد اتصال به SQL Server"""
    return pyodbc.connect(
        'DRIVER={SQL Server};'
        'SERVER=MKZ-DSAS\\DSAS;'
        'DATABASE=DSAS;'
        'UID=datadriven;'
        'PWD=5Rdx@4Rfv1362'
    )

def fetch_data_and_save_to_excel():
    """
    اجرای کوئری و ذخیره نتیجه در فایل اکسل با نام فرمت: AssetID_g13.xlsx
    """
    try:
        # برقراری اتصال به دیتابیس
        conn = get_sql_connection()
        
        # کوئری مورد نظر
        query = """
        SELECT TOP (10000000) [ID]
              ,[AssetID]
              ,[UnitID]
              ,[Value]
              ,[RecordTime]
              ,[RecordDate]
              ,[PersonelID]
              ,[OutofRange]
              ,[ValueType]
              ,[MobileID]
              ,[DateTime]
              ,[TimeStamp]
              ,[Job]
              ,[IsDeleted]
              ,[ShiftCode]
              ,[OnTime]
        FROM [DSAS].[PDA].[Periodic_Values]
        WHERE [UnitID]=21 AND
        ([AssetID]=12235)
        """
        
        # خواندن داده‌ها با pandas
        print("در حال دریافت داده‌ها از دیتابیس...")
        df = pd.read_sql(query, conn)
        
        # بستن اتصال
        conn.close()
        
        # بررسی اینکه داده‌ای وجود دارد یا خیر
        if df.empty:
            print("هیچ داده‌ای برای AssetID=12210 یافت نشد!")
            return
        
        # دریافت AssetID از دیتا (مقدار ثابت 12210 است)
        asset_id = df['AssetID'].iloc[0] if not df.empty else 12210
        
        # ساخت نام فایل با فرمت AssetID_g13.xlsx
        filename = f"{asset_id}_g13.xlsx"
        
        # ذخیره در فایل اکسل
        print(f"در حال ذخیره {len(df)} رکورد در فایل {filename}...")
        df.to_excel(filename, index=False, engine='openpyxl')
        
        print(f"✅ فایل با موفقیت ذخیره شد: {filename}")
        print(f"تعداد رکوردهای ذخیره شده: {len(df)}")
        
        # نمایش اطلاعات اولیه برای تایید
        print("\nنمونه‌ای از داده‌ها:")
        print(df.head())
        
        return filename
        
    except pyodbc.Error as e:
        print(f"❌ خطا در اتصال به دیتابیس: {e}")
        return None
    except Exception as e:
        print(f"❌ خطا: {e}")
        return None

# اجرای تابع اصلی
if __name__ == "__main__":
    fetch_data_and_save_to_excel()
```
